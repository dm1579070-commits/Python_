# Задание 2. Виселица (Угадай слово)
### Суть: Игрок угадывает скрытое слово, называя буквы. За каждую ошибку отнимается «жизнь».
Основные требования:
1. Заранее подготовленный массив/список слов. Случайный выбор одного.
2. Слово отображается в виде прочерков или нижних подчеркиваний (например: _ _ _ _ _).
3. При вводе буквы программа проверяет, есть ли она в слове. Если да — открывает все такие буквы. Если нет — вычитает жизнь.
4. Вывод сообщения о победе или поражении.


```python
import random
set=['sun', 'happy', 'peak', 'sunday']
ncomputer = random.choice (set)
emphasis=['_']*len(ncomputer)
life=6
while '_' in emphasis and life > 0:
  l=input('Буква: ') #буква
  if l in ncomputer:
    for i in range (len(ncomputer)): #проверяет каждую букву
      if ncomputer[i]==l:
        emphasis[i]=l
        print (" ".join(emphasis)) #Выводит текущий результат
  else:
      print('Попробуй еще')
      life-=1
if life>0:
 print ('Победа')
else:
   print('Поражение')
```

# Задание 3. Угадай число
### Суть: Программа загадывает число от 1 до 100. Пользователь пытается угадать это число.
Основные требования:
1. После попытки угадать программа сообщает один из вариантов:
  - Не угадал! Загаданное число БОЛЬШЕ
  - Не угадал! Загаданное число МЕНЬШЕ
  - Поздравляем ты угадал число!
2. Программа считает количество неудачных попыток и выводит счет в конце игры.
3. \* Улучшите игру так чтобы пользователь мог сам указывать диапазон загаднного числа. Диапазон не должен быть меньше 10. И не должен быть отрицательным.
4. \** Определить минимальное необходимое число попыток для нахождения загаданного числа. Если пользователь не укладывается в это количество попыток - то он проигрывает. (Задача учитывается только при выполнении пункта 3.

```python
import random
n=""
g=0
while True:
   low=int(input('Нижний диапазон: '))
   high=int(input('Верхний диапазон: '))
   if high - low <10 or high - low <0 :
    print('Диапозон меньше 10 или отрицательный')
    continue
   break

ncomputer = random.randint(low, high)

while n!=ncomputer:
   n=int (input ('Введи число: '))
   if n>ncomputer:
     print("Загаданное число МЕНЬШЕ")
     g+=1
   elif n < ncomputer:
     print ('Загаданное число БОЛЬШЕ')
     g+=1
print ('Ты угадал!')
print (f"Kол-во неудач: {g}")
```

# Задание 4. Математический спринт
### Суть: Игра на скорость. За отведенное время (например, 60 секунд) нужно решить как можно больше примеров.
Основные требования:
1. Программа генерирует случайные примеры (сложение, вычитание,
умножение).
2. Игрок вводит ответ. Если верно — засчитывается очко и дается следующий пример. Если неверно — пример повторяется или пропускается (на твой выбор).
3. В консоли должно быть видно оставшееся время (или таймер просто работает в фоне).
4. По истечении времени игра останавливается и выводит итоговый счет и статистику (сколько верных/неверных).

4 задача

```python
import random
import time
vern=0
nev=0
seconds=int(input('Сколько секунд: '))
start_time=time.time()
while True:
    set=['+', '-', '*']
    ncomputer= random.choice(set)
    x1 = random.randint(1, 10)
    x2 = random.randint(1, 10)
    if ncomputer=='+':
       exs1 = x1 + x2
    elif ncomputer=='-':
       exs2= x1 - x2
    else:
       exs3= x1 * x2

    print(f'{x1}{ncomputer}{x2}')
    answer=int(input("Ответ: "))
    if answer == exs1 or answer ==exs2 or answer==exs3:
       vern+=1
    else:
       nev+=1


    if time.time()-start_time>seconds:
       print("Время вышло =( ")
       break
    else:
       print(int(time.time()-start_time)) #Сколько осталось


print("Верных ответов: ", vern)
print("Неверных ответов: ", nev)
```
